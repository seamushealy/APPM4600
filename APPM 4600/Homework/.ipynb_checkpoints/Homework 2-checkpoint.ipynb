```python
import numpy as np
# 2
matrix = 0.5*np.array([[1, 1], 
                   [1 + (1e-10), 1 - 1e-10]])

cond_number = np.linalg.cond(matrix)/1e10

print(f"Condition Number: {cond_number}")

b1 = 1e-5
b2 = 1e-5
rel = (1/np.sqrt(2)) * np.sqrt( ((1-1e10)*b1 + b2*1e10)**2 + ( (1+1e10)*b1 - b2*1e10 )**2 )
print(rel)
```
