```python
import numpy as np
# 2
matrix = 0.5*np.array([[1, 1], 
                   [1 + (1e-10), 1 - 1e-10]])

cond_number = np.linalg.cond(matrix)/1e10

print(f"Condition Number: {cond_number}")

b1 = 1e-5
b2 = 1e-5
rel = (1/np.sqrt(2)) * np.sqrt( ((1-1e10)*b1 + b2*1e10)**2 + ( (1+1e10)*b1 - b2*1e10 )**2 )
print(rel)
```

```python
# 6
import numpy as np
import matplotlib.pylab as plt
import matplotlib
import math
%matplotlib inline
matplotlib.rcParams['mathtext.fontset'] = 'cm'
matplotlib.rcParams['font.family'] = 'STIXGeneral'

x = np.linspace(-5, 5, 1000)
fx = x - 4*np.sin(2*x) - 3

plt.plot(x, fx, label = "f(x)", color = "blue")
plt.title("Plot of f(x)")
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.savefig("Hw 2 Q6.pdf")
plt.show()
```
