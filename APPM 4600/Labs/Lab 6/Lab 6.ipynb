```python
import numpy as np
import math
import time
from numpy.linalg import inv 
from numpy.linalg import norm 

# 1 
# The condition we chose is to recompute the Jacobian every 3 steps

# 2
def LazyNewton(x0,tol,Nmax):

    ''' Lazy Newton = use only the inverse of the Jacobian for initial guess'''
    ''' inputs: x0 = initial guess, tol = tolerance, Nmax = max its'''
    ''' Outputs: xstar= approx root, ier = error message, its = num its'''

    J = evalJ(x0)
    Jinv = inv(J)
    for its in range(Nmax):

       F = evalF(x0)
       x1 = x0 - Jinv.dot(F)
       
       if (norm(x1-x0) < tol):
           xstar = x1
           ier =0
           return[xstar, ier,its]
           
       x0 = x1
    
    xstar = x1
    ier = 1
    return[xstar,ier,its]
    
def evalF(x): 
# vector function that you want to find the roots of

    F = np.zeros(2)
    
    F[0] = 4*x[0]**2 + x[1]**2 - 4
    F[1] = x[0] + x[1] - np.sin(x[0] - x[1])
 
    return F
    
def evalJ(x): 
# Jacobian of the vector function you want to find the roots of
    
    J = np.array([[8*x[0], 2*x[0]], 
                  [1 - np.cos(x[0] - x[1]), 1 + np.cos(x[0] - x[1])]])

    return J
    
def SlackerNewton(x0,tol,Nmax):

    ''' Lazy Newton = use only the inverse of the Jacobian for initial guess'''
    ''' inputs: x0 = initial guess, tol = tolerance, Nmax = max its'''
    ''' Outputs: xstar= approx root, ier = error message, its = num its'''
    ''' Recomputes Jacobian every 3 steps '''

    J = evalJ(x0)
    Jinv = inv(J)
    for its in range(Nmax):
    
       F = evalF(x0)
       x1 = x0 - Jinv.dot(F)
        
    # recompute the Jacobian
       if its % 3 == 0:
           J = evalJ(x1)
           Jinv = inv(J) 
        
       if (norm(x1-x0) < tol):
           xstar = x1
           ier =0
           return[xstar, ier,its]  
       x0 = x1
        
    xstar = x1
    ier = 1
    return[xstar,ier,its]

# 3
x0 = np.array([1, 0])
tol = 1e-10

#LazyNewton(x0, tol, 100)
SlackerNewton(x0, tol, 100)

# 4
# Both of our Slacker Newton's converge in the same number of iterations as Lazy Newton

# 5
# My Slacker Newton converges in the same number of iterations as Lazy Newton
```
