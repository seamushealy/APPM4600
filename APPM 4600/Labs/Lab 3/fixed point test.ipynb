```python
import numpy as np

def fixedpt(f, x0, tol, Nmax):
    import numpy as np
    '''x0 = initial guess'''
    '''Nmax = max number of iterations'''
    '''tol = stopping tolerance'''

    count = 0
    phat = np.array([])
    while count < Nmax:
        count = count + 1
        x1 = f(x0)
        phat = np.append(phat, x1)
        if abs(x1 - x0) < tol:
            xstar = x1
            ier = 0
            return [xstar, ier, count, phat]

        x0 = x1
        
    xstar = x1
    ier = 1
    count = 0
    return [xstar, ier, count, phat]
    
x0 = 1
tol = 1e-8
Nmax = 100

def f(x):
    return x**2 - 1

[xstar, ier, count, phat] = fixedpt(f, x0, tol, Nmax)

#print(xstar)
#print(ier)
#print(count)
#print(phat)
```
